```python
import pandas as pd


def check_clipped_features(df, threshold_pct=1.0):
  """Mendeteksi kolom yang nilainya menumpuk di nilai Max/Min (Capping/Clipping).

  :param df: DataFrame train
  :param threshold_pct: Batas minimal persentase data persis di nilai
  Max/Min (default: 1.0%)
  """
  results = []
  num_cols = df.select_dtypes(include=["number"]).columns

  for col in num_cols:
    valid_data = df[col].dropna()
    total_count = len(valid_data)
    if total_count == 0:
      continue

    max_val = valid_data.max()
    min_val = valid_data.min()

    # Hitung berapa banyak baris yang bernilai persis MAX & MIN
    max_count = (valid_data == max_val).sum()
    min_count = (valid_data == min_val).sum()

    max_pct = (max_count / total_count) * 100
    min_pct = (min_count / total_count) * 100

    if max_pct >= threshold_pct:
      results.append({
          "Column": col,
          "Type": "MAX Capped",
          "Capped Value": max_val,
          "Count": max_count,
          "Percentage (%)": round(max_pct, 2),
      })

    if min_pct >= threshold_pct:
      results.append({
          "Column": col,
          "Type": "MIN Capped",
          "Capped Value": min_val,
          "Count": min_count,
          "Percentage (%)": round(min_pct, 2),
      })

  return pd.DataFrame(results)


# --- CARA PAKAI ---
# Ganti df_train dengan nama dataframe kamu
clipped_report = check_clipped_features(train, threshold_pct=1.0)
print(clipped_report)
```
