# First Project: Building a Machine Learning Model

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

## Load Data

```python
new_data = pd.read_csv("delaney_solubility_with_descriptors.csv")
new_data
```

```python
new_data.info()
```

## Data Preparation

### Separate into x and y

```python
y = new_data['logS']
y
```

```python
x = new_data.drop('logS', axis=1)
x
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=100)
```

### **Model Building**

#### **Linear Regression**

##### **Training Model**

```python
from sklearn.linear_model import LinearRegression

model_1 = LinearRegression()

model_1.fit(X_train, y_train)
```

##### **Applying the model to make a Prediction**

```python
y_pred_model_train = model_1.predict(X_train)
y_pred_model_test = model_1.predict(X_test)
```

##### **Evaluate Model Performance**

```python
from sklearn.metrics import mean_squared_error, r2_score

train_model_1_mse = mean_squared_error(y_train, y_pred_model_train)
train_model_1_r2 = r2_score(y_train, y_pred_model_train)

test_model_1_mse = mean_squared_error(y_test, y_pred_model_test)
test_model_1_r2 = r2_score(y_test, y_pred_model_test)
```

```python
print("Model 1 MSE (Train) : ", train_model_1_mse)
print("Model 1 R2 (Train) : ", train_model_1_r2)
print("Model 1 MSE (Test) : ", test_model_1_mse)
print("Model 1 R2 (Test) : ", test_model_1_r2)
```

```python
Model_1_results = pd.DataFrame(['Linear Regression', train_model_1_mse, train_model_1_r2, test_model_1_mse, test_model_1_r2]).transpose()
Model_1_results.columns = ['Method', 'Train MSE', 'Train R2', 'Test MSE', 'Test R2']
```

```python
Model_1_results
```

#### **XGBoost and LightGBM**

```python
import xgboost as xgb
import lightgbm as lgb
import pandas as pd
from sklearn.metrics import mean_squared_error, r2_score

# 1. XGBoost Regressor
xgb_model = xgb.XGBRegressor(
    n_estimators=100, 
    max_depth=3, 
    learning_rate=0.1, 
    random_state=100
)
xgb_model.fit(X_train, y_train)
xgb_preds = xgb_model.predict(X_test)

xgb_mse = mean_squared_error(y_test, xgb_preds)
xgb_r2 = r2_score(y_test, xgb_preds)


# 2. LightGBM Regressor
lgb_model = lgb.LGBMRegressor(
    n_estimators=100, 
    max_depth=3, 
    learning_rate=0.1, 
    random_state=100,
    verbose=-1
)
lgb_model.fit(X_train, y_train)
lgb_preds = lgb_model.predict(X_test)

lgb_mse = mean_squared_error(y_test, lgb_preds)
lgb_r2 = r2_score(y_test, lgb_preds)


# 3. Print Results
results = pd.DataFrame({
    'Model': ['XGBoost', 'LightGBM'],
    'MSE': [xgb_mse, lgb_mse],
    'R2 Score': [xgb_r2, lgb_r2]
})

print("=== Model Performance Comparison ===")
print(results.to_string(index=False))
```

```python
# Test Predictions
xgb_test_preds = xgb_model.predict(X_test)

# Test Metrics
xgb_test_mse = mean_squared_error(y_test, xgb_test_preds)
xgb_test_r2 = r2_score(y_test, xgb_test_preds)

print("=== XGBoost Test Metrics ===")
print(f"Test MSE : {xgb_test_mse:.4f}")
print(f"Test R2  : {xgb_test_r2:.4f}")
```

```python
lgb_test_preds = lgb_model.predict(X_test)

# Test Metrics
lgb_test_mse = mean_squared_error(y_test, lgb_test_preds)
lgb_test_r2 = r2_score(y_test, lgb_test_preds)

print("=== LightGBM Test Metrics ===")
print(f"Test MSE : {lgb_test_mse:.4f}")
print(f"Test R2  : {lgb_test_r2:.4f}")
```

```python
Model_4_results = pd.DataFrame(['XGBoost', xgb_mse, xgb_r2, xgb_test_mse, xgb_test_r2]).transpose()
Model_4_results.columns = ['Method', 'Train MSE', 'Train R2', 'Test MSE', 'Test R2']

Model_5_results = pd.DataFrame(['LightGBM', lgb_mse, lgb_r2, lgb_test_mse, lgb_test_r2]).transpose()
Model_5_results.columns = ['Method', 'Train MSE', 'Train R2', 'Test MSE', 'Test R2']
```

#### **Looking for the best model**

```python
from lazypredict.Supervised import LazyRegressor
from sklearn.metrics import mean_squared_error
import pandas as pd

# 1. Initialize LazyRegressor with mean_squared_error as custom metric
reg = LazyRegressor(
    verbose=0,
    ignore_warnings=True,
    custom_metric=mean_squared_error
)

# 2. Fit models on your existing split data
models, _ = reg.fit(X_train, X_test, y_train, y_test)

# 3. Format custom metric column name and clean the output table
models = models.rename(columns={'mean_squared_error': 'MSE', 'R-Squared': 'R2'})

# 4. Sort models by lowest MSE and select the Top 3
top_3_models = models.sort_values(by='MSE', ascending=True).head(3)

# 5. Print the top 3 models with MSE and R2
print("=== Top 3 Best Models (Sorted by MSE) ===")
print(top_3_models[['MSE', 'R2']])
```
