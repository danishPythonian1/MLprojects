```python
missing_count = train.isnull().sum()

missing_summary = pd.DataFrame({
    "missing_count": missing_count,
    "missing_percentage": (missing_count / len(train) * 100).round(2)
}).sort_values("missing_count", ascending=False)

display(missing_summary)
```

```python
total_duplicate_rows = train.duplicated().sum()

print(f"Jumlah baris duplikat penuh: {total_duplicate_rows}")

duplicate_rows = train[train.duplicated(keep=False)].sort_values(
    by=train.columns.tolist()
)

display(duplicate_rows)
```

```python
unique_summary = pd.DataFrame({
    "unique_count": train.nunique(dropna=False),
    "dtype": train.dtypes.astype(str),
    "sample_values": [
        train[col].drop_duplicates().head(5).tolist()
        for col in train.columns
    ]
}).sort_values("unique_count")

display(unique_summary)
```

```python
# 1. Isi missing value semua kolom numerik dengan median,
#    kecuali stress_level, gender, addicted_label, dan id

excluded_cols = ["stress_level", "gender", "addicted_label", "id"]

median_cols = [
    col for col in train.columns
    if col not in excluded_cols and pd.api.types.is_numeric_dtype(train[col])
]

train[median_cols] = train[median_cols].fillna(train[median_cols].median())
```

```python
# 2. Ubah gender yang missing/NaN/null menjadi kategori "other"

train["gender"] = (
    train["gender"]
    .astype("string")
    .str.strip()
    .str.lower()
    .replace({"nan": pd.NA, "null": pd.NA, "": pd.NA})
    .fillna("other")
)
```

```python
# 3. Ubah stress_level menjadi ordinal:
#    low = 0, medium = 1, high = 2

stress_mapping = {
    "low": 0,
    "medium": 1,
    "high": 2
}

train["stress_level"] = (
    train["stress_level"]
    .astype("string")
    .str.strip()
    .str.lower()
    .replace({"nan": pd.NA, "null": pd.NA, "": pd.NA})
    .map(stress_mapping)
)
```

```python
# 4. Isi stress_level yang missing dengan nilai modus

stress_mode = train["stress_level"].mode()[0]

train["stress_level"] = (
    train["stress_level"]
    .fillna(stress_mode)
    .astype(int)
)
```

```python
# Ubah missing/NaN/null pada academic_work_impact menjadi kategori "other"

train["academic_work_impact"] = (
    train["academic_work_impact"]
    .astype("string")
    .str.strip()
    .str.lower()
    .replace({"nan": pd.NA, "null": pd.NA, "": pd.NA})
    .fillna("other")
)
```

```python
missing_count = train.isnull().sum()

missing_summary = pd.DataFrame({
    "missing_count": missing_count,
    "missing_percentage": (missing_count / len(train) * 100).round(2)
}).sort_values("missing_count", ascending=False)

display(missing_summary)
```
