```python
import pandas as pd
import matplotlib.pyplot as plt

train = pd.read_csv("train_FE1.csv")
```

```python
train.columns
```

```python
train.shape
```

```python
train_sample = train.sample(n=10000, random_state=42)

plt.figure(figsize=(10,6))
plt.scatter(train_sample['gaming_hours'], train_sample['addicted_label'])
plt.xlabel('gaming hours')
plt.ylabel('addicted label')
plt.title('gaming vs addicted')
plt.grid(alpha=0.3)

plt.show()
```
